<h1 style = "color : teal"> One Hot Encoding </h1>
One-Hot Encoding (OHE) creates a binary column for every distinct category present in the nominal feature

```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv('cars.csv')
```

```python
df.sample(6)
```

```python
# checking the counts of each brand type 
df['owner'].value_counts()
```

<h1 style = "color : teal">One Hot Encoding using Pandas</h1>

```python
# get_dummies function helps to do one hot encoding using pandas 
pd.get_dummies(df, columns = ['owner', 'fuel'], dtype = int)

# df is the dataframe and columns attribute holds the list of columns to be encoded else all columns are encoded
# by default the datatype returned is boolean so dtype is used to transoform 
```

<h1 style = "color : teal">Drop the first column </h1>


```python
# dropping the first column because,
# This creates perfect multicollinearity—a condition where one input feature can be predicted linearly from the remaining input features

pd.get_dummies(df, columns = ['owner', 'fuel'], drop_first = True, dtype = int)
```

<h1 style = "color : teal">Using Scikit Learn 

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 0:4], df.iloc[:, -1], test_size = 0.2, random_state = 2)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(sparse_output = False, dtype = np.int32, drop = 'first')

# the OneHotEncoder class returns a sparse matrix so we pass sparse_output = False to get an numpy array
# the np arrya is float so dtype helps to get required datatype
# drop attribute helps to drop the first column from all the features that are using this class 
```

```python
ohe.fit_transform(X_train[['fuel', 'owner']])
```

```python
ohe.transform(X_test[['fuel', 'owner']])
```

```python
X_train_new = ohe.fit_transform(X_train[['fuel', 'owner']])
X_test_new = ohe.transform(X_test[['fuel', 'owner']])
```

```python
# adding the encoded columns with other columns
np.hstack((X_train[['brand','km_driven']].values,X_train_new))
```
